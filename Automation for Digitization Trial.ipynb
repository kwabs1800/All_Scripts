```python
import arcpy, sys
aprx=arcpy.mp.ArcGISProject("Current")
mp=aprx.activeMap
if mp is None:
    print("No Active Map Set")
    sys.exit()
arcpy.env.workspace=r"C:\Users\micha\Downloads\Digitization\Digitization_Trial.gdb" #set the existing GDB as your workspace

#layers = mp.listLayers() #layers from active map
#if not layers:
#    print("No Layers Found")
#    sys.exit()
#for lyr in layers:
#    print(lyr.name) #test to see if the layers in the map show up

parcels_lyr = mp.listLayers("Parcels")
if not parcels_lyr:
    print("Parcels layer not found")
    sys.exit()
parcels_lyr = parcels_lyr[0] #select and use parcels layer for field calculations

update_calc= arcpy.management.SelectLayerByAttribute(parcels_lyr, "NEW_SELECTION", "Area_Ha IS NULL")
arcpy.management.CalculateFields( update_calc, "PYTHON3", [["Area_Ha", "!shape.area! * 0.000009290304"], ["Area_Acres", "!shape.area!*0.00002295682"]]) #calculate fields by converting from square feet to acres and hectares

#export parcels layer to dwg format in Output folder
output_type = "DWG_R2018"
output_file = r"C:\Users\micha\Downloads\Digitization\Outputs\Parcels.dwg"
arcpy.conversion.ExportCAD([update_calc], output_type, output_file)

#export parcels layer to shape files in Output folder
Output_Folder = r"C:\Users\micha\Downloads\Digitization\Outputs"
arcpy.conversion.FeatureClassToShapefile([update_calc], Output_Folder)
```
